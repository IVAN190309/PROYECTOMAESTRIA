# Optimizacion de costos operativos
### El siguiente proyecto, tiene por objetivo optimizar el costo operativo de una ruta ya establecida que correspondea un circuito, se pretende hacer una comparativa con una actualizaciín de tarifas para así poder tomar la decisión en si la operacion sigue siendo rentable o no.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==============================================================================
# ESTRUCTURA METODOLÓGICA DEL PROYECTO FINAL APLICATIVO
# ==============================================================================
print("==================================================================")
print(" OPTIMIZACIÓN DE RUTAS Y COSTOS LOGÍSTICOS")
print(" Maestría en Logística Internacional y Cadena de Suministros")
print("==================================================================\n")

# ------------------------------------------------------------------------------
# 1. IDENTIFICACIÓN Y DESCRIPCIÓN DEL PROBLEMA & 2. DEFINICIÓN DEL OBJETIVO
# ------------------------------------------------------------------------------
print("--- 1. IDENTIFICACIÓN, DESCRIPCIÓN Y OBJETIVO ---")
print("• Contexto: Operación de transporte de carga en el corredor León - Silao - Vallejo.")
print("• Problemática: Ineficiencia en tiempos de tránsito, consumo de combustible y costos operativos globales.")
print("• Objetivo: Cuantificar y aplicar un modelo de optimización de transporte que reduzca los costos operativos globales y los tiempos de entrega de la flota.\n")

# ------------------------------------------------------------------------------
# 3. INFORMACIÓN Y DATOS 
# ------------------------------------------------------------------------------
print("--- 3. INFORMACIÓN Y DATOS (MATRICES DE ENTRADA) ---")

# Datos del Escenario Actual
escenario_actual = [
    {"No": 1, "Origen": "LEÓN GTO", "Destino": "SILAO GTO", "KM": 24, "Tiempo": 0.50, "Costo_Comb": 27.0, "Total_Comb": 259.2, "Rendimiento": 2.5, "Peajes": 0.0, "Mantenimiento_km": 25.2},
    {"No": 2, "Origen": "SILAO GTO", "Destino": "VALLEJO MX", "KM": 355, "Tiempo": 4.15, "Costo_Comb": 27.0, "Total_Comb": 3834.0, "Rendimiento": 2.5, "Peajes": 2163.0, "Mantenimiento_km": 532.5},
    {"No": 3, "Origen": "VALLEJO MX", "Destino": "SILAO GTO", "KM": 355, "Tiempo": 4.15, "Costo_Comb": 27.0, "Total_Comb": 3834.0, "Rendimiento": 2.5, "Peajes": 2357.0, "Mantenimiento_km": 532.5},
    {"No": 4, "Origen": "SILAO GTO", "Destino": "LEÓN GTO", "KM": 24, "Tiempo": 0.50, "Costo_Comb": 27.0, "Total_Comb": 259.2, "Rendimiento": 2.5, "Peajes": 0.0, "Mantenimiento_km": 25.2}
]

# Datos del Escenario a Optimizar
escenario_optimizar = [
    {"No": 1, "Origen": "LEÓN GTO", "Destino": "SILAO GTO", "KM": 24, "Tiempo": 0.30, "Costo_Comb": 27.0, "Total_Comb": 231.43, "Rendimiento": 2.8, "Peajes": 0.0, "Mantenimiento_km": 25.2},
    {"No": 2, "Origen": "SILAO GTO", "Destino": "VALLEJO MX", "KM": 355, "Tiempo": 4.15, "Costo_Comb": 27.0, "Total_Comb": 3423.21, "Rendimiento": 2.8, "Peajes": 2163.0, "Mantenimiento_km": 532.5},
    {"No": 3, "Origen": "VALLEJO MX", "Destino": "SILAO GTO", "KM": 355, "Tiempo": 4.15, "Costo_Comb": 27.0, "Total_Comb": 3423.21, "Rendimiento": 2.8, "Peajes": 2163.0, "Mantenimiento_km": 532.5},
    {"No": 4, "Origen": "SILAO GTO", "Destino": "LEÓN GTO", "KM": 24, "Tiempo": 0.30, "Costo_Comb": 27.0, "Total_Comb": 231.43, "Rendimiento": 2.8, "Peajes": 0.0, "Mantenimiento_km": 25.2}
]

df_actual = pd.DataFrame(escenario_actual)
df_opt = pd.DataFrame(escenario_optimizar)

print("\n> Tabla 1: Escenario Actual Base")
print(df_actual.to_string(index=False))

print("\n> Tabla 2: Escenario a Optimizar")
print(df_opt.to_string(index=False))

# ------------------------------------------------------------------------------
# 4. SELECCIÓN DEL MÉTODO CUANTITATIVO & 5. APLICACIÓN DEL MODELO
# ------------------------------------------------------------------------------
# "Modelo de Transporte / Programación Lineal"
def calcular_costos_totales(df):
    total_km = df['KM'].sum()
    total_tiempo = df['Tiempo'].sum()
    total_combustible = df['Total_Comb'].sum()
    total_peajes = df['Peajes'].sum()
    total_mantenimiento = df['Mantenimiento_km'].sum() # Sumatoria total del costo de mantenimiento por tramo
    costo_global = total_combustible + total_peajes + total_mantenimiento
    return total_km, total_tiempo, total_combustible, total_peajes, total_mantenimiento, costo_global

km_act, t_act, comb_act, peajes_act, mant_act, costo_tot_act = calcular_costos_totales(df_actual)
km_opt, t_opt, comb_opt, peajes_opt, mant_opt, costo_tot_opt = calcular_costos_totales(df_opt)

# ------------------------------------------------------------------------------
# 6. ANÁLISIS DE RESULTADOS Y TABLAS COMPARATIVAS
# ------------------------------------------------------------------------------
print("\n--- 6. ANÁLISIS DE RESULTADOS ---")
ahorro_pesos = costo_tot_act - costo_tot_opt
ahorro_porcentaje = (ahorro_pesos / costo_tot_act) * 100

df_comparativo = pd.DataFrame({
    'Indicador Clave de Desempeño': ['Kilómetros Totales', 'Tiempo Estimado Total', 'Costo Combustible', 'Costo Peajes', 'Costo Mantenimiento', 'Costo Operativo Global'],
    'Escenario Actual': [f"{km_act} km", f"{t_act:.2f} hrs", f"${comb_act:,.2f} MXN", f"${peajes_act:,.2f} MXN", f"${mant_act:,.2f} MXN", f"${costo_tot_act:,.2f} MXN"],
    'Escenario Optimizado': [f"{km_opt} km", f"{t_opt:.2f} hrs", f"${comb_opt:,.2f} MXN", f"${peajes_opt:,.2f} MXN", f"${mant_opt:,.2f} MXN", f"${costo_tot_opt:,.2f} MXN"]
})

print(df_comparativo.to_string(index=False))
print(f"\n[Resultado Clave] Ahorro Financiero Total Obtenido: ${ahorro_pesos:,.2f} MXN ({ahorro_porcentaje:.2f}% de reducción).")

# ------------------------------------------------------------------------------
# GENERACIÓN DE GRÁFICA
# ------------------------------------------------------------------------------
plt.figure(figsize=(10, 5))
categorias = ['Combustible', 'Peajes', 'Mantenimiento', 'Costo Global Operativo']
valores_actual = [comb_act, peajes_act, mant_act, costo_tot_act]
valores_opt = [comb_opt, peajes_opt, mant_opt, costo_tot_opt]

x = np.arange(len(categorias))
width = 0.35

plt.bar(x - width/2, valores_actual, width, label='Escenario Actual', color='#d62728')
plt.bar(x + width/2, valores_opt, width, label='Escenario Optimizado', color='#1f77b4')

plt.ylabel('Costo en MXN', fontsize=10, fontweight='bold')
plt.title('Comparativa Metodológica: Escenario Actual vs. Escenario Optimizado', fontsize=11, fontweight='bold')
plt.xticks(x, categorias, fontsize=10)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()

plt.savefig('analisis_comparativo_costos.png', dpi=300)
plt.show()
print("\n[Éxito] Gráfica generada y guardada como 'analisis_comparativo_costos.png'.")

# ------------------------------------------------------------------------------
# 7. PROPUESTA DE SOLUCIÓN & 8. CONCLUSIONES
# ------------------------------------------------------------------------------
print("\n--- 7. PROPUESTA DE SOLUCIÓN Y CONCLUSIONES ---")
print("• Propuesta: Institucionalizar la mejora en rendimiento de flota (de 2.5 a 2.8 km/l) y la renegociación de peajes de retorno en las rutas hacia Vallejo MX.")
print("• Conclusiones: El modelo cuantitativo de transporte demuestra que optimizar el rendimiento y los tiempos de tránsito genera un impacto financiero directo y mejora significativamente la eficiencia de la cadena de suministro.")
print("==================================================================")
```
